```python
import pandas as pd
import numpy as np
from pathlib import Path
from utils import detrend_ue
OUT_DIR = Path("results")
OUT_DIR.mkdir(parents=True, exist_ok=True)


methods_dict = {
    'MaximumSequenceProbability': 'MSP',
    'Perplexity': 'PPL',
    'MeanTokenEntropy': 'MTE',
    'MonteCarloSequenceEntropy': 'MCSE',
    'MonteCarloNormalizedSequenceEntropy': 'MCNSE',
    'LexicalSimilarity_rougeL': 'LSRL',
}

MODELS = {
    'llama': 'Llama 3.1 8B',
    'gemma': 'Gemma 2 9B',
    'eurollm': 'EuroLLM 9B',
}

DATASETS = {
    'metricx-metricx-24-hybrid-xxl-v2p6': [
        'wmt14_csen',
        'wmt14_deen',
        'wmt14_ruen',
        'wmt14_fren',
        'wmt19_deen',
        'wmt19_fien',
        'wmt19_lten',
        'wmt19_ruen',
    ],
    'XComet-XCOMET-XXL': [
        'wmt14_csen',
        'wmt14_deen',
        'wmt14_ruen',
        'wmt14_fren',
        'wmt19_deen',
        'wmt19_fien',
        'wmt19_lten',
        'wmt19_ruen',
    ],
    'Comet-wmt22-comet-da': [
        'wmt14_csen',
        'wmt14_deen',
        'wmt14_ruen',
        'wmt14_fren',
        'wmt19_deen',
        'wmt19_fien',
        'wmt19_lten',
        'wmt19_ruen',
    ],
    'bleu_proper': [
        'wmt14_csen',
        'wmt14_deen',
        'wmt14_ruen',
        'wmt14_fren',
        'wmt19_deen',
        'wmt19_fien',
        'wmt19_lten',
        'wmt19_ruen',
    ],
}

METRICS = {
    'metricx-metricx-24-hybrid-xxl-v2p6': 'MetricX XXL',
    'XComet-XCOMET-XXL': 'XComet XXL',
    'Comet-wmt22-comet-da': 'Comet WMT22',
}

def write_metric_csvs_all_models():
    """
    For each metric, create ONE CSV with all models:
      - columns: ['model', 'method', <datasets...>]
      - rows: for each model and method, both raw and detrended (method-LINE)
    """
    for metric, metric_pretty in METRICS.items():
        datasets = DATASETS[metric]
        ue_methods_short = list(methods_dict.values())

        rows = []
        for model_key, model_name in MODELS.items():
            # get UE scores (raw + detrended) for the current model
            ue_scores, _, _ = detrend_ue(
                datasets=datasets,
                model=model_key,
                model_type="base",
                all_metrics=[metric],
                ue_methods=ue_methods_short,
                methods_dict=methods_dict,
                return_unprocessed=True,
            )

            # build rows for this model
            for m_short in ue_methods_short:
                raw_vals = np.asarray(ue_scores[f"{m_short}_raw"], dtype=float)
                detr_vals = np.asarray(ue_scores[f"{m_short}_detr"], dtype=float)

                rows.append(
                    {"model": model_key, "method": m_short, **{ds: v for ds, v in zip(datasets, raw_vals)}}
                )
                rows.append(
                    {"model": model_key, "method": f"{m_short}-LINE", **{ds: v for ds, v in zip(datasets, detr_vals)}}
                )

        # assemble and write the CSV for this metric
        df = pd.DataFrame(rows, columns=["model", "method"] + datasets)

        # Optional niceties:
        # df = df.round(2)  # keep full precision if you prefer
        # sort rows by model then method
        df = df.sort_values(by=["model", "method"]).reset_index(drop=True)

        out_path = OUT_DIR / f"{metric}.csv"
        df.to_csv(out_path, index=False)
        print(f"wrote {out_path}")
        

if __name__ == "__main__":
    write_metric_csvs_all_models()
```

```python
import pandas as pd
import numpy as np
from pathlib import Path
from utils import detrend_ue_w_quality

OUT_DIR = Path("results")
OUT_DIR.mkdir(parents=True, exist_ok=True)

SELECT_MODELS = {
    "llama": MODELS["llama"],
    "gemma": MODELS["gemma"],
}

SELECT_METRICS = {
    "AlignScoreInputOutput": "Align Score",
    "Accuracy": "Acc",
}
DATASETS = {
        'AlignScoreInputOutput' : ['xsum'],
        'Accuracy' : ['gsm8k'],
}

def write_selected_metric_csvs_all_models():
    """
    For each selected metric, create ONE CSV with llama & gemma:
      columns: ['model','method', <datasets...>]
      rows: for each model and method, both raw and detrended (method-LINE)
    Also writes a tidy (long) version alongside.
    """
    ue_methods_short = list(methods_dict.values())

    for metric_key, metric_pretty in SELECT_METRICS.items():
        datasets = DATASETS[metric_key]
        rows = []

        for model_key, _model_name in SELECT_MODELS.items():
            # raw + detrended UE scores aligned to datasets
            ue_scores, _, _, _, _ = detrend_ue_w_quality(
                datasets=datasets,
                model=model_key,
                model_type="base",
                all_metrics=[metric_key],
                ue_methods=ue_methods_short,
                methods_dict=methods_dict,
                return_unprocessed=True,
            )

            for m_short in ue_methods_short:
                raw_vals = np.asarray(ue_scores[f"{m_short}_raw"], dtype=float)
                detr_vals = np.asarray(ue_scores[f"{m_short}_detr"], dtype=float)

                rows.append({"model": model_key, "method": m_short,        **{ds: v for ds, v in zip(datasets, raw_vals)}})
                rows.append({"model": model_key, "method": f"{m_short}-LINE", **{ds: v for ds, v in zip(datasets, detr_vals)}})

        # wide format
        df = pd.DataFrame(rows, columns=["model", "method"] + datasets)
        df = df.sort_values(by=["model", "method"]).reset_index(drop=True)
        out_wide = OUT_DIR / f"{metric_key}.csv"  # e.g., AlignScoreInputOutput.csv, Accuracy.csv
        df.to_csv(out_wide, index=False)

        # tidy/long format (optional, handy for plotting/analysis)
        df_long = df.melt(id_vars=["model", "method"], var_name="dataset", value_name="score")
        out_long = OUT_DIR / f"{metric_key}__long.csv"
        df_long.to_csv(out_long, index=False)

        print(f"Wrote {out_wide} and {out_long}")

if __name__ == "__main__":
    write_selected_metric_csvs_all_models()
```
